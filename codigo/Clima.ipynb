```python
# prompt: convertir a un dataframe

import pandas as pd
import requests


# Tu clave API de OpenWeather
api_key = 'COLOCAR API KEY'

# URL base de la API de OpenWeather
base_url = 'http://api.openweathermap.org/data/2.5/weather'

# Lista de ciudades peruanas
ciudades_peru = [
    'Amazonas', 'Ancash', 'Apurímac', 'Arequipa', 'Ayacucho', 'Cajamarca', 'Cusco', 'Huancavelica', 'Huánuco', 'Ica', 'Junín', 'La Libertad', 'Lambayeque', 'Lima', 'Loreto', 'Madre de Dios', 'Moquegua', 'Pasco', 'Piura', 'Puno', 'San Martín', 'Tacna', 'Tumbes', 'Ucayali'
]

data = []

for ciudad in ciudades_peru:
    params = {
        'q': f'{ciudad},pe',
        'appid': api_key,
        'units': 'metric',
        'lang': 'es'
    }
    respuesta = requests.get(base_url, params=params)

    if respuesta.status_code == 200:
        datos = respuesta.json()
        data.append({
            'Ciudad': ciudad,
            'Temperatura': datos['main']['temp'],
            'Descripción': datos['weather'][0]['description'],
            'Humedad': datos['main']['humidity'],
            'Viento': datos['wind']['speed']
        })
    else:
        print(f"Error: No se pudo obtener el clima para {ciudad}. Código de error: {respuesta.status_code}")

df = pd.DataFrame(data)
df


```

```python
#Metricas de Calidad
exitos = 0
fallos = 0

if respuesta.status_code == 200:
    exitos += 1
else:
    fallos += 1

print(f"Ciudades procesadas correctamente: {exitos}")
print(f"Fallos de conexión o datos: {fallos}")


#validar el procentaje de valores duplicados

duplicated_percentage = (df.duplicated().sum() / len(df)) * 100
print(f"Porcentaje de Valores Duplicados: {duplicated_percentage:.2f}%")

```
